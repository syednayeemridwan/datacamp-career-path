# Terminologies

- POJO: 
    - plain old java objects. 
    - simple classes that store and manage data. 
    - getters and setters are key ingredients in POJOs
    - should be public classes, but the fields are private
    - Should have default/no-argument constructor
    - Don't extend/implement other class
    - Do not contain business logic
- Wrapper classes: 
    - helps us to treat primitive types like objects, like putting them in a collection
    - have additional methods and fields, static methods and fields
    - Allows `null` like actual object null values
- Packages: 
    - Organized code of related classes
    - built-in packages start with `java` or `javax`. and custom packages starts with reverse organization name, something like `com.`
    - packages use periods (.) to delimit name parts 
    - Needs to `import` the package before starting to use it
    - `BigInteger`, `BigDecimal` are not wrapper classes, they are classes of java's `math` package
