# Terminologies

- POJO: 
    - plain old java objects. 
    - simple classes that store and manage data. 
    - getters and setters are key ingredients in POJOs
    - should be public classes, but the fields are private
    - Should have default/no-argument constructor
    - Don't extend/implement other class
    - Do not contain business logic
- Wrapper classes: 
    - helps us to treat primitive types like objects, like putting them in a collection
    - have additional methods and fields, static methods and fields
    - Allows `null` like actual object null values
- Packages: 
    - Organized code of related classes
    - built-in packages start with `java` or `javax`. and custom packages starts with reverse organization name, something like `com.`
    - packages use periods (.) to delimit name parts 
    - Needs to `import` the package before starting to use it
    - `BigInteger`, `BigDecimal` are not wrapper classes, they are classes of java's `math` package
- Java have 2 types:
    - Primitive type
    - Reference type (This is what we understand when type is called)
- Collections:
    - A package of types to store and manipulate a group of objects
    - this data structure works like array- except it is dynamically sized, stores only objects, heterogeneous objects allowed, use methods for object retrieval
    - List: Ordered list of objects, allow duplicates
    - Set: Un-ordered list of objects. unique elements. it is an interface.
    - Queue: FIFO list of objects. `ArrayBlockingQueue` from `java.util.concurrent`
    - Map: Look-up type. has key-value pair (both key and value are objects). No null key allowed.
    - Generic: Used to specify what type of objects are stored in collections with a `<>`. It is enforced so that type safety is complied.
    - Autoboxing: Process of automatically putting primitives into a wrapper class when inserting inside collections.
- Exceptions:
    - Happens when something goes wrong. It is recoverable.
    - Try block isolates the code that produces exception
    - Catch block handles the exceptions
    - One try block may include multiple catch blocks
    - finally block occurs whether a catch block executes or not. It contains clean-up or tear-down code.
- Error:
    - Serious issue. It is not recoverable
    - It makes the program crash
    - Runtime Exception.
    - Programming mistake. good programming required to avoid these exceptions
- Dynamically loading a class: `Class myClass = Class.forName("com.mysql.Driver");`
- Throws:
    - Another way to catch an exception
    - Use throws on any method to pass the exception back to the calling method during calling time. It helps passing the responsibility for handling the exception to a caller instead of inside the method itself
    - Helps to centralize the exceptions and reduce try-catch block
    - Chaining of throwing exceptions is re-throwing the exceptions
    - can also include `throw new Exception();`
- Logging:
    - Logging assists developers monitor and understand application activity for debugging
    - Uses multiple levels to put emphasis/importance on the logged information. this is what differentiate it from `println` command.
    - Logs can be recorded in a separate file
    - Built-in logging class
    - `Logger` class from `java.util.logging.*;`
    - When initiating the Logger object, specify the class name it is in `public static Logger logger = Logger.getLogger("HelloWorld");`

# Most Important Interview Conversions

```java
// Array -> List
List<String> list = Arrays.asList(arr);

// List -> Array
String[] arr = list.toArray(new String[0]);

// List -> Set
Set<String> set = new HashSet<>(list);

// Set -> List
List<String> list = new ArrayList<>(set);

// Map Keys -> List
List<String> keys = new ArrayList<>(map.keySet());

// Map Values -> List
List<Integer> values = new ArrayList<>(map.values());

// String -> List
List<String> list = Arrays.asList(str.split(","));

// List -> String
String str = String.join(",", list);

// String -> int
int n = Integer.parseInt(str);

// int -> String
String str = String.valueOf(n);

// Object -> String
String s = obj.toString();

// String -> char[]
char[] chars = str.toCharArray();

// char[] -> String
String str = new String(chars);
```

These conversions alone cover roughly 80–90% of the collection and string conversion questions commonly encountered in Java interviews and day-to-day development.

# Exception Handling Patterns (Most Common)

### Basic

```java
try {
    save();
} catch (IOException e) {
    logger.error("Save failed", e);
}
```

### Multiple Catch

```java
try {
    process();
}
catch (IOException e) {
    logger.error("IO error", e);
}
catch (SQLException e) {
    logger.error("DB error", e);
}
```

### Multi-Catch

```java
catch (IOException | SQLException e) {
    logger.error("Processing failed", e);
}
```

### Custom Exception

```java
public class UserNotFoundException extends RuntimeException {
    public UserNotFoundException(String msg) {
        super(msg);
    }
}
```

### Exception Wrapping

```java
catch (SQLException e) {
    throw new RuntimeException("Database failed", e);
}
```
