# Terminology

- class: a container for organizing Java code
- method: Methods as reusable blocks of code
- Control flow: if,else, else if and switch (switch does not work with double or float)

# Methods

- Void returns nothing and performs an action

```python
public class HelloWorld {
public static void main (String[] args) {
    double smallTip = computeTip(55.0, 0.20); //compute a 20 percent tip on 55 dollars, Calling the method
    double bigTip = computeTip(135.50, 0.33); //compute a 33 percent tip on 135 dollars, Calling the method
    System.out.println("Small tip: $" + smallTip);
    System.out.println("Nice tip: $" + bigTip);
}

// Define the method
static double computeTip (double billAmount, double tipPercentage) {
        double tip = billAmount * tipPercentage;
        return tip;
    }
}
```

# Control Flow (`If-Else`) with Logical Expressions

```python
int a = 14;

if (a > 10) && (a < 20) {
    System.out.println ("Size is average");
} else if (a == 10) || (a == 20) {
    System.out.println ("Size is on the border");
} else {
    System.out.println ("Size is abnormal");
}
```

# Switch : Alternative to If-Else

```python
char direction = 'Z';
switch (direction) {
    case 'E':
        System.out.println("Go east");
        break;
    case 'W':
        System.out.println("Go west");
        break;
    case 'N':
        System.out.println("Go north");
        break;
    case 'S':
        System.out.println("Go south");
        break;
    default: // else alternative that works in switch
        System.out.println("We are lost");
}
```

# For loop

```python
// For loop
for (int i = 0; i < 10; i++){
    System.out.println("Loop number: " + i);
}

// For each loop (normally used array to take from the beginning to the end)
int[] prices = {6, 4, 3, 7};
int sum = 0;
for(int price: prices){
    sum = sum + price;
}
System.out.println(sum);
```

# While Loop

```python
// While loop
int i = 5;
while (i < 100) {
    i = i * 2;
    System.out.println("Looping ...");
}

// Do-While loop (Even if the conditions are not met, the loop will be executed at least once)
int i = 5;
do {
    i = i * 2;
    System.out.println("Looping...");
} while (i < 100);
```

# Exception Handling

```python
int x = 10;
int y = 0;
try {
    System.out.println(x/y);
} catch (ArithmeticException ex){
    System.out.println("Divide by zero?");
} catch (Exception e){ // chaining multiple exceptions
    System.out.println("Some other exception");
} finally {
    System.out.println("Clean up here");
}
```
