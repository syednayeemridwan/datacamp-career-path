# Cheat-Sheet

- FileReader and FileWriter reads and writes by characters
- Iterator helps to traverse collections: `Iterator<String> it = col.iterator();`
- ListIterator only works for list: `ListIterator<String> it = names.listIterator();`
    - But it has backward capability: `ListIterator<String> it = textHistory.listIterator(textHistory.size());`
- Stream: 
    - Improves performance for large data. It process Collections functionally
    - `Stream<String> stream = col.stream();` then `stream.forEach(name -> System.out.println(name));`
    - Helps to filter, map and transform data
    - Streams are lazy pipelines, not data stores. So you need Collectors otherwise the result will vanish: `resultStream.collect(Collectors.toSet())`

Below is a **clean, interview-ready + practical cheat sheet** for the three requested Java areas.

# 📁 Java File I/O Classes (Reader / Writer / Streams)

These are part of **legacy IO (java.io)** but still widely used in interviews and basic programs.

# 1. FileReader, FileWriter, BufferedReader, BufferedWriter

## 📄 FileReader (Character-based input)

| Method       | Purpose          | One-liner Example    |
| ------------ | ---------------- | -------------------- |
| read()       | Read single char | `int c = fr.read();` |
| read(char[]) | Read into buffer | `fr.read(buffer);`   |
| close()      | Close stream     | `fr.close();`        |

### Example

```java id="fr1"
FileReader fr = new FileReader("a.txt");
int c;
while ((c = fr.read()) != -1) {
    System.out.print((char)c);
}
fr.close();
```

---

## 📝 FileWriter (Character-based output)

| Method        | Purpose      | Example              |
| ------------- | ------------ | -------------------- |
| write(int)    | Write char   | `fw.write('A');`     |
| write(String) | Write text   | `fw.write("Hello");` |
| write(char[]) | Write array  | `fw.write(arr);`     |
| flush()       | Force write  | `fw.flush();`        |
| close()       | Close stream | `fw.close();`        |

### Example

```java id="fw1"
FileWriter fw = new FileWriter("a.txt");
fw.write("Hello Java");
fw.flush();
fw.close();
```

---

# ⚡ BufferedReader (FAST reading)

Wraps FileReader → improves performance.

| Method     | Purpose         | Example          |
| ---------- | --------------- | ---------------- |
| readLine() | Read full line  | `br.readLine();` |
| read()     | Read char       | `br.read();`     |
| close()    | Close stream    | `br.close();`    |
| lines()    | Stream of lines | `br.lines()`     |

### Example (VERY IMPORTANT)

```java id="br1"
BufferedReader br = new BufferedReader(new FileReader("a.txt"));

String line;
while ((line = br.readLine()) != null) {
    System.out.println(line);
}

br.close();
```

---

# ✍️ BufferedWriter (FAST writing)

| Method    | Purpose      | Example              |
| --------- | ------------ | -------------------- |
| write()   | Write text   | `bw.write("Hello");` |
| newLine() | New line     | `bw.newLine();`      |
| flush()   | Flush buffer | `bw.flush();`        |
| close()   | Close stream | `bw.close();`        |

### Example

```java id="bw1"
BufferedWriter bw = new BufferedWriter(new FileWriter("a.txt"));

bw.write("Hello World");
bw.newLine();
bw.write("Java IO");
bw.close();
```

---

# 🔄 KEY CONCEPT (VERY IMPORTANT)

| Class          | Type       | Use                       |
| -------------- | ---------- | ------------------------- |
| FileReader     | Low-level  | Read characters           |
| FileWriter     | Low-level  | Write characters          |
| BufferedReader | High-level | Fast reading + readLine() |
| BufferedWriter | High-level | Fast writing              |

---

# ⚡ COMBINED REAL-WORLD PATTERN

```java id="combo1"
BufferedReader br = new BufferedReader(new FileReader("in.txt"));
BufferedWriter bw = new BufferedWriter(new FileWriter("out.txt"));

String line;
while ((line = br.readLine()) != null) {
    bw.write(line.toUpperCase());
    bw.newLine();
}

br.close();
bw.close();
```

---

# 🔄 CONVERSIONS (IMPORTANT INTERVIEW POINT)

| Conversion              | Example                      |
| ----------------------- | ---------------------------- |
| File → Reader           | `new FileReader(file)`       |
| File → Writer           | `new FileWriter(file)`       |
| Reader → BufferedReader | `new BufferedReader(reader)` |
| Writer → BufferedWriter | `new BufferedWriter(writer)` |
| BufferedReader → Stream | `br.lines()`                 |
| String → File output    | `fw.write(str)`              |

---

# ⚡ MODERN ALTERNATIVE (VERY IMPORTANT)

Java 8+ prefers **NIO Files API** instead of these:

| Old IO         | Modern Equivalent                        |
| -------------- | ---------------------------------------- |
| FileReader     | `Files.newBufferedReader(path)`          |
| FileWriter     | `Files.newBufferedWriter(path)`          |
| BufferedReader | `Files.readAllLines()` / `Files.lines()` |
| BufferedWriter | `Files.write()`                          |

### Example

```java id="nio1"
List<String> lines = Files.readAllLines(Paths.get("a.txt"));
Files.write(Paths.get("b.txt"), lines);
```
